Discretization

EQUAL WIDTH DISCRETISATION

```python
equal_width_df = mean_df.copy()

num_bins = 10

columns_to_discretize = ['norm_LotFrontage', 'norm_GarageYrBlt']

for column_name in columns_to_discretize:
    column_values = equal_width_df[column_name]
    min_value = column_values.min()
    max_value = column_values.max()

    bin_width = (max_value - min_value) / num_bins

    bins = [min_value + i * bin_width for i in range(num_bins + 1)]

    new_column_name = f'equal_width_{column_name}'
    equal_width_df[new_column_name] = pd.cut(column_values, bins=bins, precision=2)

print(equal_width_df[['equal_width_norm_LotFrontage', 'equal_width_norm_GarageYrBlt']].head())
```

```python
import pandas as pd
from sklearn.preprocessing import KBinsDiscretizer
import matplotlib.pyplot as plt

k_df = mean_df

attributes_to_discretize = ['LotFrontage', 'GarageYrBlt']
num_bins = 10

fig, axes = plt.subplots(1, len(attributes_to_discretize), figsize=(10, 4))

for idx, attribute in enumerate(attributes_to_discretize):
    discretization = KBinsDiscretizer(n_bins=num_bins, encode='ordinal', strategy='kmeans')
    discretization.fit(k_df[[attribute]])
    intervals = discretization.bin_edges_[0].tolist()


    bin_labels = [f'Bin_no_{i}' for i in range(1, num_bins + 1)]

    ax = axes[idx]
    ax.hist(k_df[attribute], bins=intervals, edgecolor='k')
    ax.set_xlabel(attribute)
    ax.set_ylabel('Frequency')
    ax.set_title(f'Histogram with {num_bins} Bins for {attribute}')

    ax.set_xticks([(intervals[i] + intervals[i+1]) / 2 for i in range(num_bins)])
    ax.set_xticklabels(bin_labels, rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

OUTLIER TRIMMING

```python
import seaborn as sns
import matplotlib.pyplot as plt

attributes = ['norm_LotFrontage','norm_GarageYrBlt']

fig, axes = plt.subplots(1, 2, figsize=(10, 2))

for i, attribute in enumerate(attributes):


    Q1 = mean_df[attribute].quantile(0.25)
    Q3 = mean_df[attribute].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.2* IQR
    upper_bound = Q3 + 1.0 * IQR


    trimmed_df = mean_df[(mean_df[attribute] >= lower_bound) & (mean_df[attribute] <= upper_bound)]
    sns.boxplot(x=trimmed_df[attribute], ax=axes[i])
    axes[i].set_title(f'Trimmed Box Plot of {attribute}')
    axes[i].set_xlabel(attribute)

plt.tight_layout()
plt.show()
```

OUTLIER CAPPING

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

cap_df = mean_df.copy()

attributes = ['norm_LotFrontage','norm_GarageYrBlt']

fig, axes = plt.subplots(1, 2, figsize=(10, 2))

for i, attribute in enumerate(attributes):

    lower_cap = cap_df[attribute].quantile(0.05)
    upper_cap = cap_df[attribute].quantile(0.95)


    capped_df = cap_df.copy()
    capped_df[attribute] = capped_df[attribute].clip(lower=lower_cap, upper=upper_cap)


    sns.boxplot(x=capped_df[attribute], ax=axes[i])
    axes[i].set_title(f'Capped Box Plot of {attribute}')
    axes[i].set_xlabel(attribute)

plt.tight_layout()
plt.show()
```
